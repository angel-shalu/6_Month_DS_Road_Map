```python
dataset = pd.read_csv(r"C:\Users\shali\Desktop\DS_Road_Map\12. Deep Learning\ANN\TENSORFLOW KERAGS\Churn_Modelling.csv")
X = dataset.iloc[:, 3:-1].values
y = dataset.iloc[:, -1].values

print("X Sample:\n", X[:3])
print("y Sample:\n", y[:10])
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Label Encoding Gender
le = LabelEncoder()
X[:, 2] = le.fit_transform(X[:, 2])

# One Hot Encoding Geography
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])], remainder='passthrough')
X = np.array(ct.fit_transform(X))

print("After Encoding:\n", X[:3])
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
X = sc.fit_transform(X)

print("After Scaling:\n", X[:3])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)

print("Train Shape:", X_train.shape)
print("Test Shape:", X_test.shape)
print("Train Shape:", y_train.shape)
print("Test Shape:", y_test.shape)
```

```python
# Initializing the ANN
ann = tf.keras.models.Sequential()     #establish layer one after another
ann.add(tf.keras.layers.Dense(units=6))    #input layer

# Input + First Hidden Layer
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))      #hidden layer1

# Second Hidden Layer
ann.add(tf.keras.layers.Dense(units=5, activation='relu'))      #hidden layer2

# Third Hidden Layer
ann.add(tf.keras.layers.Dense(units=4, activation='relu'))      #hidden layer3

# Output Layer
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

```python
# Compile ANN
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
# Train ANN
history = ann.fit(X_train, y_train, batch_size = 32, epochs = 100)
```
